```python
import json
def pretty_print(title, data):
    print("\n" + "="*20 + f" {title} " + "="*20)
    print(json.dumps(data, indent=2))

try:
    # 1. Profile
    profile = kite.profile()
    pretty_print("PROFILE", profile)

    # 2. Margins / Funds
    margins = kite.margins()
    pretty_print("MARGINS", margins)

    # 3. Holdings
    holdings = kite.holdings()
    pretty_print("HOLDINGS", holdings)

    # 4. Positions
    positions = kite.positions()
    pretty_print("POSITIONS", positions)

    # 5. Orders list
    orders = kite.orders()
    pretty_print("ORDERS", orders)

    # 6. Trades (for all completed orders)
    trades = kite.trades()
    pretty_print("TRADES", trades)

    # 7. Mutual fund holdings
    mf_holdings = kite.mf_holdings()
    pretty_print("MF HOLDINGS", mf_holdings)

    # 8. Mutual fund orders
    mf_orders = kite.mf_orders()
    pretty_print("MF ORDERS", mf_orders)

except Exception as e:
    print("❌ Error fetching data:", e)
```
